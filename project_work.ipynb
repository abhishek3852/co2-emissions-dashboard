```python
for column in a.columns:
    if a[column].isna().mean() * 100 > 70:
        a.drop(columns=[column], inplace=True)
```

```python
a = a[a["iso_code"].notna()]
```

```python
# Q1 — Which 10 countries emitted the most CO₂ in 2023, and how has their share of global emissions changed since 1990?

countries_high = (
    a[a["year"] == 2023]
    .groupby("country")["co2"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

countries_past = (
    a[
        (a["year"] == 1990) &
        (a["country"].isin(countries_high.index))
    ]
)[["country", "share_global_co2"]]

# Plotly Visualization

fig = px.bar(
    countries_high.reset_index(),
    x="co2",
    y="country",
    orientation="h",
    title="Top 10 CO₂ Emitting Countries in 2023",
    labels={
        "co2": "CO₂ Emissions (Million Tonnes)",
        "country": "Country"
    },
    template="plotly_white",
    text="co2",
    height=600
)

fig.update_yaxes(autorange="reversed")

fig.update_traces(
    texttemplate="%{text:.0f}",
    textposition="outside"
)

fig.update_layout(
    title_x=0.5,
    xaxis_title="CO₂ Emissions (Million Tonnes)",
    yaxis_title="Country"
)

fig.show()
```

```python
# Q2 — Which countries have the highest CO₂ emissions per capita in 2023?
q2 = (
    a[a["year"] == 2023]
    .sort_values("co2_per_capita", ascending=False)
    [["country","co2_per_capita"]]
    .head(10)
)

fig = px.bar(
    q2,
    x="co2_per_capita",
    y="country",
    orientation="h",
    title="Top 10 Countries by CO₂ Per Capita (2023)",
    template="plotly_white"
)

fig.update_yaxes(autorange="reversed")
fig.show()
```

```python
# Q3 — How have global CO₂ emissions changed since 1990?

q3 = (
    a[a["year"] >= 1990]
    .groupby("year")["co2"]
    .sum()
    .reset_index()
)

fig = px.line(
    q3,
    x="year",
    y="co2",
    title="Global CO₂ Emissions Since 1990",
    template="plotly_white"
)

fig.show()
```

```python
# Q4 — Which countries reduced CO₂ emissions the most between 1990 and 2023?

co2_1990 = a[a["year"]==1990][["country","co2"]].rename(columns={"co2":"co2_1990"})
co2_2023 = a[a["year"]==2023][["country","co2"]].rename(columns={"co2":"co2_2023"})

q4 = pd.merge(co2_1990, co2_2023, on="country")

q4["reduction"] = q4["co2_1990"] - q4["co2_2023"]

q4 = q4.sort_values("reduction", ascending=False).head(10)

fig = px.bar(
    q4,
    x="reduction",
    y="country",
    orientation="h",
    title="Countries with Largest CO₂ Reduction (1990–2023)",
    template="plotly_white"
)

fig.update_yaxes(autorange="reversed")
fig.show()
```

```python
# Q5 — Is there a relationship between population and CO₂ emissions in 2023?

q5 = a[a["year"]==2023][["country","population","co2"]].dropna()

fig = px.scatter(
    q5,
    x="population",
    y="co2",
    hover_name="country",
    title="Population vs CO₂ Emissions (2023)",
    template="plotly_white"
)

fig.show()
```

```python
# Q6 — Relationship between GDP and CO₂ (2022)

q6 = a[a["year"]==2022][["country","gdp","co2"]].dropna()

fig = px.scatter(
    q6,
    x="gdp",
    y="co2",
    hover_name="country",
    title="GDP vs CO₂ Emissions (2022)",
    template="plotly_white"
)

fig.show()
```

```python
# Q7 — G7 CO₂ Trend Since 1990

g7 = [
    "Canada",
    "France",
    "Germany",
    "Italy",
    "Japan",
    "United Kingdom",
    "United States"
]

q7 = a[
    (a["country"].isin(g7)) &
    (a["year"]>=1990)
]

fig = px.line(
    q7,
    x="year",
    y="co2",
    color="country",
    title="G7 CO₂ Emission Trends Since 1990",
    template="plotly_white"
)

fig.show()
```

```python
#Q8 — Coal, Oil and Gas CO₂ Since 1990

q8 = (
    a[a["year"]>=1990]
    .groupby("year")[["coal_co2","oil_co2","gas_co2"]]
    .sum()
    .reset_index()
)

q8 = q8.melt(
    id_vars="year",
    var_name="Fuel",
    value_name="CO₂"
)

fig = px.line(
    q8,
    x="year",
    y="CO₂",
    color="Fuel",
    title="Global CO₂ by Fuel Type Since 1990",
    template="plotly_white"
)

fig.show()
```

```python
# Q9 — Which countries experienced the largest increase in CO₂ emissions between 1990 and 2023?

co2_1990 = a[a["year"] == 1990][["country","co2"]].rename(columns={"co2":"co2_1990"})
co2_2023 = a[a["year"] == 2023][["country","co2"]].rename(columns={"co2":"co2_2023"})

q9 = pd.merge(co2_1990, co2_2023, on="country")

q9["Increase"] = q9["co2_2023"] - q9["co2_1990"]

q9 = q9.sort_values("Increase", ascending=False).head(10)

fig = px.bar(
    q9,
    x="Increase",
    y="country",
    orientation="h",
    title="Top 10 Countries with the Largest CO₂ Increase (1990–2023)",
    template="plotly_white"
)

fig.update_yaxes(autorange="reversed")
fig.show()
```

```python
#Q10 — Which countries have the highest CO₂ emissions per unit of GDP? (2022)


q10 = (
    a[a["year"] == 2022]
    .dropna(subset=["co2_per_gdp"])
    .sort_values("co2_per_gdp", ascending=False)
    [["country","co2_per_gdp"]]
    .head(10)
)

fig = px.bar(
    q10,
    x="co2_per_gdp",
    y="country",
    orientation="h",
    title="Top 10 Countries by CO₂ per GDP (2022)",
    template="plotly_white"
)

fig.update_yaxes(autorange="reversed")
fig.show()
```
